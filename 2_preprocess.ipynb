# 2. Predprocesiranje podataka

Ovaj notebook provodi čišćenje izvornog dataseta:
1. Uklanjanje stupaca s >90% NULL vrijednosti
2. Standardizacija naziva stupaca (lowercase, underscore)
3. Provjera duplikata
4. Podjela na trening (80%) i test (20%) skup

```python
import pandas as pd

CSV_FILE_PATH = 'Support_tickets_named.csv'
df = pd.read_csv(CSV_FILE_PATH, delimiter=',')
print(f"Izvorni dataset: {df.shape[0]} redaka, {df.shape[1]} stupaca")
```

## 2.1 Uklanjanje stupaca s previše NULL vrijednosti

```python
# Prag: 90% praznih vrijednosti
threshold = 0.9

# Prikaži što se briše
for col in df.columns:
    null_pct = df[col].isnull().mean()
    if null_pct >= threshold:
        print(f"  BRIŠEM: {col:40s} ({null_pct*100:.1f}% NULL)")

df = df.loc[:, df.isnull().mean() < threshold]

print(f"\nNakon čišćenja: {df.shape[0]} redaka, {df.shape[1]} stupaca")
```

## 2.2 Standardizacija naziva stupaca

```python
df.columns = df.columns.str.lower().str.replace(' ', '_')
print(f"Stupci: {list(df.columns)}")
```

## 2.3 Provjera duplikata

```python
duplicates = df.duplicated().sum()
print(f"Broj duplikata: {duplicates}")
```

## 2.4 Spremanje čistog dataseta

```python
df.to_csv('cleaned_data.csv', index=False)
print(f"Spremljeno: cleaned_data.csv ({df.shape[0]} redaka)")
```

## 2.5 Podjela na trening (80%) i test (20%) skup

```python
df20 = df.sample(frac=0.2, random_state=1)
df80 = df.drop(df20.index)

df80.to_csv('Support_tickets_PROCESSED.csv', index=False)
df20.to_csv('Support_tickets_PROCESSED_20.csv', index=False)

print(f"Trening skup (80%): {df80.shape[0]} redaka -> Support_tickets_PROCESSED.csv")
print(f"Test skup (20%):    {df20.shape[0]} redaka -> Support_tickets_PROCESSED_20.csv")
```
