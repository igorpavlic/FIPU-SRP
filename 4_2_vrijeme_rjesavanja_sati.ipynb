```python
import pandas as pd

# 1. Učitavanje i ispravna pretvorba datuma (korištenjem ISO8601 formata)
df['created_at'] = pd.to_datetime(df['created_at'], format='ISO8601')
df['resolved_at'] = pd.to_datetime(df['resolved_at'], format='ISO8601')

# 2. Izračun razlike u satima
# .dt.total_seconds() vraća razliku u sekundama, koju dijelimo s 3600 za sate
df['vrijeme_rjesavanja_sati'] = (df['resolved_at'] - df['created_at']).dt.total_seconds() / 3600

# 3. Dodatna obrada (opcionalno)
# Postavljanje negativnih vrijednosti na 0 ili NULL (u slučaju grešaka u podacima)
df.loc[df['vrijeme_rjesavanja_sati'] < 0, 'vrijeme_rjesavanja_sati'] = 0

# Zaokruživanje na dvije decimale radi preglednosti u skladištu
df['vrijeme_rjesavanja_sati'] = df['vrijeme_rjesavanja_sati'].round(2)

print("Metrika 'vrijeme_rjesavanja_sati' je uspješno izračunata.")
```
