# 1. Osnovna analiza podataka

Ovaj notebook provodi početnu analizu izvornog skupa podataka `Support_tickets_named.csv`:
- Dimenzije dataseta, tipovi podataka
- Nedostajuće vrijednosti (NULL)
- Broj unikatnih vrijednosti po stupcu
- Distribucija ključnih kategorijskih varijabli

Pitanja:
- Je li je skup podataka dovoljno velik? -- 66691 redaka, 58 stupaca
- Je li skup ima dovoljno različite podatke? -- uključuje različite kategorije poput prioriteta, statusa i drugih atributa koji opisuju zahtjeve
- Je li skup ima vremensku dimenziju? -- sadrži podatke vezane uz vrijeme, poput datuma kreiranja ili rješavanja zahtjeva, što omogućuje analizu kroz vrijeme
- Je li skup ima kvantitativne i kvalitativne podatke? -- sadrži i kvantitativne (vrijeme rješavanja) i kvalitativne podatke (statusa ili prioriteta)
- Je li skup ima puno nedostajućih vrijednosti? -- prisutan je veći broj nedostajućih vrijednosti, što je zahtijevalo dodatnu obradu

```python
import pandas as pd

CSV_FILE_PATH = 'Support_tickets_named.csv'
df = pd.read_csv(CSV_FILE_PATH, delimiter=',')
print(f"Dataset učitan: {df.shape[0]} redaka, {df.shape[1]} stupaca")
print(f"\nPrvih 5 redaka:")
df.head()
```

## 1.1 Tipovi podataka

```python
print(df.dtypes)
```

## 1.2 Nedostajuće vrijednosti (NULL)

```python
null_counts = df.isnull().sum()
null_pct = (df.isnull().mean() * 100).round(1)
null_df = pd.DataFrame({'NULL count': null_counts, 'NULL %': null_pct})
print(null_df[null_df['NULL count'] > 0].sort_values('NULL %', ascending=False))
```

## 1.3 Broj unikatnih vrijednosti

```python
print(df.nunique())
```

## 1.4 Distribucija kategorijskih varijabli

```python
kategorijski = ['issue_proj', 'issue_type', 'issue_priority', 'issue_status', 'issue_resolution']
for col in kategorijski:
    print(f"\n=== {col} ===")
    print(df[col].value_counts())
```

## 1.5 Statistika numeričkih varijabli

```python
print(df[['wf_total_time', 'processing_steps', 'issue_comments_count']].describe())
```
